## Strings

```python

my_basic_string = "sympathy"

my_long_string = "Roses are red, violets are blue, \"What comes first? egg or chicken?\", \"Sorry sir, I don\'t have a clue."

# Unicode is the default string format in Python 3.
my_direct_unicode_string = "Unicode: ₹, अ, 😀"

my_pi = "\u03C0 = 3.14\U0001F600"

# Multi line Strings
kauffman_poem = """
I have folded my sorrows into the mantle of summer night,
Assigning each brief storm its allotted space in time,
Quietly pursuing catastrophic histories buried in my eyes.
"""
```


| Glyph | UTF Code | Description |

```python
# Print a string to the stdout
print(my_basic_string)
print(my_long_string)
print(my_direct_unicode_string)
print(my_pi)
print(kauffman_poem)
```

```python
# String Methods
my_basic_string.capitalize() # Return a copy of the string with its first character capitalized and the rest lowercased.
my_basic_string.lower() # Return a copy of the string with all the characters  lowercased.
my_basic_string.upper() # Return a copy of the string with all characters converted to uppercase.
my_long_string.title() # Return a titlecased version of the string where words start with an uppercase character.

my_basic_string.count('y') # Return the number of non-overlapping occurrences of substring.
my_long_string.split(" ")



my_basic_string.endswith('thy') # Returns a boolean if the end substring matches.
my_basic_string.startswith('symp') # Returns a boolean on matching the first chars of string object.

# str.format
# Basic Formatting of output using str.format() method
'{0}, {1}, {2}'.format('a', 'b', 'c')
'{}, {}, {}'.format('a', 'b', 'c') 
'{2}, {1}, {0}'.format(*'abc')  # unpacking argument sequence which we cover in a later topic


'Coordinates: {latitude}, {longitude}'.format(latitude='37.24N', longitude='-115.81W')
coord = {'latitude': '37.24N', 'longitude': '-115.81W'}
'Coordinates: {latitude}, {longitude}'.format(**coord)  # unpacking argument sequence which we cover in a later topic

# Precision
yes_votes = 42_572_654
total_votes = 85_705_149
percentage = yes_votes / total_votes
'{:-9} YES votes  {:2.2%}'.format(yes_votes, percentage)

# Alignments: left->{:<n}, right -> {:>n} and center -> {:^n}
'{:<30}'.format("alignment")
'{:>30}'.format("alignment")
'{:^30}'.format("alignment")


# Fill chars {:A^n}
'{:*<30}'.format("alignment")
'{:+>30}'.format("alignment")
'{:_^30}'.format("alignment")

# Accessing arguments’ items
mountains = ["Nilgiris", "Vindhyas", "Aravallis", "Satpuras", "Ghats", "Shivaliks"]
months = ["January", "February", "March", "April", "May", "June", "July",
          "August", "September", "October", "November", "December"]
'This year I will be hiking through the {} in the month of {}.'.format(mountains[2], months[4])





print(kauffman_poem.splitlines()) # Return a list of the lines in the string, breaking at line boundaries. 
## This is different from
print(kauffman_poem.split('\n'))

```
