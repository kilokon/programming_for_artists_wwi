## Binary Representation of Numbers
- Bits: The fundamental unit of information in a computer, represented as either 0 or 1.
- Byte: A group of 8 bits (common unit for storing characters).
- Word: A word refers to the number of bits a processor can handle at a time. This size can vary depending on the architecture (e.g., 32-bit or 64-bit processors).
- Nibbles: A group of 4 bits.


Binary Grouping	| Size	| Example	| Decimal Equivalent|
|:---|:--|:--|:--|
Bit|	1 bit	|1|	1
Nibble|	4 bits|	1001|	9
Byte	|8 bits|	01101001	|105
Word (platform-dependent)|	16 or 32 bits|	00000000 00001111|	15

### Hexadecimal Numbers

### Step by step conversion from binary to hex to decimal

| Part          | Value     | Explanation                            |
|---------------|-----------|----------------------------------------|
| Binary        | `10101011`| 8 bits = 1 byte                        |
| Nibble 1      | `1010`    | Left 4 bits → hex `A` = decimal `10`   |
| Nibble 2      | `1011`    | Right 4 bits → hex `B` = decimal `11`  |
| Hex Value     | `0xAB`    | Compact hexadecimal representation     |
| Decimal Value | `171`     | Decimal equivalent of `0xAB`           |

| Binary        | Nibble 1 | Nibble 2 | Hex | Decimal |
|---------------|----------|----------|-----|---------|
| `10101011`    | `1010`   | `1011`   | AB  | 171     |
| `01100001`    | `0110`   | `0001`   | 61  | 97      |
| `11110000`    | `1111`   | `0000`   | F0  | 240     |


```python
# Variable Declarations
foo:int = 20
bar:float = 34.243

# Binary Numbered Variables
alpha:int = 0b0101000
beta:int = 0b0101001
gamma:int = 0b0101011
# Python doesn't directly assign floating-point numbers in their binary representation to variables. 
# Workaround for floating point binary variable declaration
binary_string = "01000000100000000000000000000000"  # (Example binary representation)
number = float.fromhex("0x" + binary_string)  # Convert from hex for demonstration


# Printing Values Out in Decimal format
print(alpha)
print(beta)
print(gamma)
print(f"Number: {number}")

# Printing Values Out in Binary Format
print(f'{bin(alpha)[2:]}')
print(f'{bin(beta)[2:]}')
print(f'{bin(gamma)[2:]}')
# print(f'{bin(number)[2:]}') ## TypeError: 'float' object cannot be interpreted as an integer
```

### Two's Complement Negation
When we negate x using the unary minus sign (-), it converts x to its two's complement representation. The two's complement of a number has its bits flipped from the least significant set bit(inclusive) to the left, and the rightmost bit remains the same.

```python
# Two's Complement Negation
x:int = 233
print(f'Binary interpretation of x: {bin(x)[2:]}')

# Time to check the binary interpretation of -x
print(f'Binary interpretation of -x: -{bin(-x)[3:]}')
```
