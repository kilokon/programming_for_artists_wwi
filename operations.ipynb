## Bitwise Operations

| Operation | Result |
|---:|:---|
| `x \| y` | bitwise _or_ of _x_ and _y_ |
| `x ^ y` | bitwise _exclusive or_ of _x_ and _y_ | 
| `x & y` | bitwise _and_ of _x_ and _y_ | 
| `x << n` | _x_ shifted left by _n_ bits | 
| `x >> n` | _x_ shifted right by _n_ bits | 
| `~x` | the bits of _x_ inverted | 

**Bitwise AND (&):** This operation performs a logical AND on corresponding bits of two operands.

````python
# Example 01 
# 2 => 010
# 7 => 111
#----------
#      010
#----------
print(2&7)

# Example 02 
# 31 => 0011111
# 67 => 1000011
#--------------
#       0000011
#--------------
print(31&67)
````

**Bitwise OR (|):** This operation performs a logical OR on corresponding bits. The result is 1 if either or both bits at that position are 1. The result is 0 only if both bits are 0.

````python
# Example 01 
# 2 => 010
# 7 => 111
#----------
#      111
#----------
print(2|7)

# Example 02 
# 31 => 0011111
# 67 => 1000011
#--------------
#       1011111
#--------------
print(31|67)
````

**Bitwise XOR (^) (exclusive OR):** This operation performs a logical XOR on corresponding bits. The result is 1 if the bits at that position are different. The result is 0 if the bits are the same.

````python
# Example 01 
# 2 => 010
# 7 => 111
#----------
#      111
#----------
print(2|7)

# Example 02 
# 31 => 0011111
# 67 => 1000011
#--------------
#       1011111
#--------------
print(31|67)
````

**Bitwise NOT (~):** This is a unary operation that flips the bits of a single operand. So, a 0 becomes 1 and vice versa.

**Left shift (<<):** This operation shifts the bits of the operand to the left by the number of positions specified by the second operand. Zeros are filled in on the right side.

````python
# Example 01 
# 2 => 010
# 7 => 111
#----------
#      111
#----------
print(2|7)

# Example 02 
# 31 => 0011111
# 67 => 1000011
#--------------
#       1011111
#--------------
print(31|67)
````

**Right shift (>>):** This operation shifts the bits of the operand to the right by the number of positions specified by the second operand. The behavior on the left side depends on the programming language (signed vs unsigned shift).

````python
x = 10  # 1010 in binary
y = 6   # 0110 in binary

# AND operation
result = x & y  # 0010 (only the rightmost set bit remains)

# OR operation
result = x | y  # 1110 (all set bits from both operands)

# XOR operation
result = x ^ y  # 1100 (bits that differ between x and y)

# NOT operation (flips all bits in x)
result = ~x  # 0101

# Left shift (shifts x 2 bits to the left)
result = x << 2  # 40 (101000 in binary)

# Right shift (arithmetic shift)
result = x >> 1  # 5 (0010 in binary)
````

### Manipulating Rightmost Bits
Using the following formulas perform binary arithmatic for performance.

**Turn off the rightmost set bit.**<br>
*Example:*<br>
01011000 -> 01010000<br>
*Formula:*<br>
```
x&(x-1)
```
*Use:*<br>
- To determine if an unsigned integer is a power of 2 or 0 (apply formula then test if 0).

**Turn on the rightmost 0-bit.**<br>
*Example:*<br>
10100111 -> 10101111<br>
*Formula:*<br>
```
x|(x+1)
```
*Use:*<br>
- sfd

**Turn off all trailing set bits.**<br>
*Example:*<br>
10100111 -> 1010000
*Formula:*<br>
``` 
x&(x+1)
```
*Use:*<br>
- To determine if an unsigned integer is of form 2^n - 1,0 or all 1's.

**Turn on all trailing 0's.**<br>
*Example:*<br>
10101000 -> 10101111<br>
*Formula:*<br>
```
x|(x-1)
```
*Use:*
- sfd

**Create word with a single set bit at the position of the rightmost unset bit.**<br>
*Example:*<br>
10100111 -> 00001000<br>
*Formula:*<br>
```
~x&(x+1)
```
*Use:*
- Isolating the rightmost set bit.
- This expression is useful in algorithms to detect the lowest unset bit position in x that turns into a set bit in x + 1.

````python
x = 167  # Binary: 10100111
result = ~x & (x + 1)  # Binary: 0000 (Rightmost set bit cleared)
print(f"Original: {x} (binary: {bin(x)[2:]})")
print(f"Cleared rightmost set bit: {result} (binary: {bin(result)[2:]})")
````

**Create word with a single unsetbit at the position of the rightmost set bit.**<br>

01011000 -> 00000111<br>

*Use:*

- Can be used to fill all bits with 1 up to and including the least significant 1-bit.
- This technique can be helpful for creating masks in bit manipulation tasks.
